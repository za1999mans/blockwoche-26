1. Create a list called `movies` containing at least 4 of your favorite movie titles.

````python
movies = ["The Shawshank Redemption", "The Godfather", "The Dark Knight", "Pulp Fiction"]
movies.append("The Lord of the Rings: The Return of the King")
print(movies)
movies[1] = "Movie XY"
print(movies)
print(movies[0:2])

print("\n---Loop without index---")
for movie in movies:
    print(movie)

print("\n---Loop with index---")
for index, movie in enumerate(movies):
    print(f"{index + 1}: {movie}")
````

## Exercise 7: Student Database

Create a small program that manages a list of students.

1. Create a list called `classroom` containing at least 3 dictionaries. Each dictionary should have `"name"`, `"age"`, and `"grade"`.
2. Add a 4th student to the list.
3. Loop through the `classroom` and print: `"[Name] is [Age] years old and has grade [Grade]."`.
4. Find the student with the highest grade and print their name.
5. Convert the entire `classroom` list to a pretty JSON string and print it.

````python
# 1. Create a list called `classroom` containing at least 3 dictionaries. Each dictionary should have `"name"`, `"age"`, and `"grade"`.
classroom = [
    {"name": "Alex", "age": 16, "grade": 90},
    {"name": "Markus", "age": 15, "grade": 95},
    {"name": "Luka", "age": 15, "grade": 95}
]

# 2. Add a 4th student to the list.
classroom.append({"name": "Andreas", "age": 15, "grade": 100})

# 3. Loop through the `classroom` and print: `"[Name] is [Age] years old and has grade [Grade]."`.
for student in classroom:
    print(f"{student["name"]} is {student.get("age")} years old and has grade {student.get("grade")}.")

# 4. Find the student with the highest grade and print their name.
print("\n-------\n 4. Find the student with the highest grade and print their name.")
top_student = classroom[0]
for student in classroom:
    # if current student has a greater grade then top_student! then replace top_student
    if student.get("grade") > top_student.get("grade"):
        top_student = student

print(top_student)

````
