## `while`-loops

A `while`-loop repeats as long as a condition is true

- Structure:
```
while condition:
  code_to_repeat        # this code should make sure that the condition will be false
```

````python
number = 5

while number >= 1:
    print(number)
    number = number - 1
````

````python
number = 5

while number >= 1:
    print(number)
    number -= 1 # number = number - 1
````

````python
number = 5

while number <= 100:
    print(number)
    number += 10 # number = number + 10
````

````python
number = 5

while number <= 100:
    print(number)
    number *= 5 # number = number * 5 
````

### `break`
`break` stops a loop immediately

````python
number = 5

while number <= 100:
    print(number)
    number += 10 # number = number + 10
    if (number == 55):
        break
````

````python
number = 5

while True:
    print(number)
    number += 10 # number = number + 10
    if (number == 55):
        break
````

### `continue"
`continue` skips the rest of the current loop iteration and moves to the next one! 

````python
number = 5

while number <= 100:
    number += 10 # number = number + 10
    if number == 55:
        continue
    print(number)
    
````

## Nested Loops

A loop can be inside another loop

````python
for row in range(1,4):
    for col in range(1,4):
        print(f"row={row}, col={col}")
````

````python
for row in range(1,4):
    print(f"{row}:", end=" ")
    for col in range(1,4):
        print(f"{row * col}", end="   ")
    print()
    
````

# Checking Membership with `in`

- The `in` operator checks whether a value exists in a collection

````python
names = ["Alex", "Andreas", "Markus"]

# Check whether Andreas exists in the list names! 
for name in names:
    if name == "andreas":
        print("True")

# with the operator `in` we can avoid using the for loop
print("Andreas" in names)
print("andreas" in names)

person_dict = {"name": "Andreasa", "age": 55}
print("name" in person_dict)
print("city" in person_dict)
````
